```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier

# 1. Historical training data (Features: Amount, Distance_from_Home_km, Hour_of_Day)
X_train = pd.DataFrame([
    [15.50, 2, 14],    # Normal lunch purchase
    [4500.00, 850, 3], # Late-night, far away, high amount
    [60.00, 5, 18],    # Normal grocery run
    [3200.00, 1200, 2] # Suspicious overnight purchase
], columns=["amount", "distance_km", "hour"])

y_train = [0, 1, 0, 1]  # 0 = Legitimate, 1 = Fraud

# 2. Train the model offline
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)

# 3. Real-time inference on a single incoming transaction
live_transaction = pd.DataFrame([[2950.00, 900, 2]], columns=["amount", "distance_km", "hour"])

prediction = rf_model.predict(live_transaction)[0]
fraud_probability = rf_model.predict_proba(live_transaction)[0][1]

print(f"Decision: {'BLOCK (Fraud)' if prediction == 1 else 'APPROVE'}")
print(f"Fraud Confidence: {fraud_probability * 100:.1f}%")
```
